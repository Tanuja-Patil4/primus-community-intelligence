```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
BASE_DIR = Path.cwd().parent
DATA_DIR = BASE_DIR / "data"

communities = pd.read_csv(DATA_DIR / "communities.csv")
residents = pd.read_csv(DATA_DIR / "residents.csv")
engagement = pd.read_csv(DATA_DIR / "resident_engagement.csv")

print("Communities:", communities.shape)
print("Residents:", residents.shape)
print("Engagement:", engagement.shape)
```

```python
display(communities.head())
display(residents.head())
display(engagement.head())
```

```python
print("Missing values:")
print(engagement.isnull().sum())

print("\nDuplicate rows:")
print(engagement.duplicated().sum())

print("\nUnique residents:")
print(engagement["resident_id"].nunique())

print("\nUnique communities:")
print(engagement["community_id"].nunique())
```

```python
engagement[
    [
        "activities_attended",
        "unique_activity_types",
        "service_requests",
        "satisfaction_score"
    ]
].describe()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    engagement["satisfaction_score"],
    bins=20,
    kde=True
)

plt.title("Distribution of Resident Satisfaction")
plt.xlabel("Satisfaction Score")
plt.ylabel("Number of Records")

plt.show()
```
